# Lab 13 — Work session and peer review
*Week 13. Baseline: sections 1–4. No powerups this week.*

You have a full draft of your capstone: a brief, an app, an evaluated AI analysis, and the notebook behind
them. Tonight one classmate reads it the way a decision-maker and a grader would, and writes down what is
missing. You do the same for them. Next week you present, so this is the last cheap moment to find a problem.

This lab has no code and no course setup cell. It is a form. The reviewer fills it in; the author answers it.

may be asked to explain any of it. Cells marked *(edit this cell)* are yours to fill in.

review: the point is what *you* noticed. A review that could have been written without opening the notebook will be marked revise.

**3b. Against the rubric (35 minutes).** Now read everything. For each part of the final rubric, mark one

the grader will look.
