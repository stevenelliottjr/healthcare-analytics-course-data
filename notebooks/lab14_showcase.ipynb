## 1. The presentation: 10 minutes, hard stop
A decision-maker gives you ten minutes. Use them the way the brief is written: answer first, evidence
second. Present from your **app** (the view-only link from week 8) and your brief; slides are optional and at most five.

| Time | Part | What it must contain |
|---:|---|---|
| 0:00–1:00 | **The question** | Who decides what, and why it matters now. One sentence of context, not a literature review. |
| 1:00–3:00 | **The answer** | Your headline finding with its number, and what you recommend. If someone leaves at 3:00 they should know it. |
| 3:00–6:00 | **The app, live** | Change one input and show what moves. Narrate what a reader should look at. |
| 6:00–7:30 | **How you know** | The data and its denominator, what was suppressed or missing, the method and the baseline it beats, the evaluation. |
| 7:30–9:00 | **The AI analyst** | Its score on your evaluation set, one question it got wrong, and how you caught it. |
| 9:00–10:00 | **Limits and next step** | What this cannot tell you, and the one thing you would do with another month. |

Then **5 minutes of questions**, faculty first. Expect "explain this line of code" at least once.

**Rules that keep the evening on time**
- The timekeeper holds up a card at 8:00 and stops you at 10:00, mid-sentence if needed.
- Test screen sharing in the breakout room before the session. Have the app link and a PDF export of the
  brief open in tabs as a backup.
- Every number on screen has its denominator or its unit. Every chart title states its point.
- Rehearse once, out loud, with a timer. Most first runs are 14 minutes.

**While others present** you will fill in a one-line form per talk (the instructor posts it): the point in
one sentence, and one question you would ask. It counts toward participation.
